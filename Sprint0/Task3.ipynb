Описание задачи:
Создать базовый класс Project в C#, который будет представлять проекты в рамках
организации. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Project:
• Атрибуты: ID проекта (ProjectId), Название проекта (ProjectName), Статус
проекта (Status).
• Методы:
o UpdateStatus(): метод для обновления статуса проекта.
o GetProjectDetails(): метод для получения деталей проекта.
o AssignTeamMember(): метод для назначения командного члена
проекту.

Требования к производным классам:
1. ПродуктовыйПроект (ProductProject): Должен содержать дополнительные
атрибуты, такие как Ожидаемая дата завершения (ExpectedCompletionDate).
Метод UpdateStatus() должен быть переопределен для включения
ожидаемой даты завершения в статус проекта.
2. ИсследовательскийПроект (ResearchProject): Должен содержать
дополнительные атрибуты, такие как Финансирование (FundingAmount).
Метод AssignTeamMember() должен быть переопределен для указания
специфических требований к навыкам командного члена для
исследовательского проекта.
3. ИнфраструктурныйПроект (InfrastructureProject) (если требуется третий
класс): Должен содержать дополнительные атрибуты, такие как Срок
выполнения (ExecutionPeriod). Метод GetProjectDetails() должен быть
переопределен для отображения срока выполнения проекта вместе с
другими деталями проекта.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
class Project {
    public int ProjectId {get; set;}
    public string ProjectName {get; set;}
    public string Status {get; set;}

    protected List<string> TeamMembers = new List<string>();

    public virtual void UpdateStatus(string newStatus) {
        Status = newStatus;
        Console.WriteLine($"Статус проекта {ProjectName} обновлён на {Status}");
    }

    public virtual void GetProjectDetails() {
        Console.WriteLine($"ID проекта: {ProjectId}");
        Console.WriteLine($"Название проекта: {ProjectName}");
        Console.WriteLine($"Статус проекта: {Status}");
    }

    public virtual void AssignTeamMember(bool hasExpTeam, decimal finalAmount, string memberName, bool hasExp) {
        TeamMembers.Add(memberName);
        Console.WriteLine($"Участник {memberName} прошёл по требованиям на проект {ProjectName}");
    }
}

class ProductProject : Project {
    public DateTime ExpectedCompletionDate {get; set;}

    public override void UpdateStatus(string newStatus) {
        Status = $"{newStatus}. Ожидаемая дата завершения: {ExpectedCompletionDate:dd.MM.yyyy}";
        Console.WriteLine($"Статус проекта: {Status}");
    }
}

class ResearchProject : Project {
    public decimal FundingAmount {get; set;}

    public override void AssignTeamMember(bool hasExpTeam, decimal finalAmount, string memberName, bool hasExp) {
        Console.WriteLine($"Проверка участника {memberName} на специфические требования к навыкам для проекта {ProjectName}.");
        Console.WriteLine("Требования: опыт работы над проектами, опыт работы в команде, сумма в пределах финансирования.");
        Console.WriteLine($"Сумма финансирования: {FundingAmount}");

        if (hasExp && hasExpTeam && (finalAmount <= FundingAmount)) {
            TeamMembers.Add(memberName);
            Console.WriteLine($"Участник {memberName} подходит под требования к специфическим навыкам.");
        }
        else {
            Console.WriteLine($"Участник {memberName} не подходит под требования к специфическим навыкам.");

            if (!hasExp) {      
                Console.WriteLine("Нет опыта.");
            }
            if (!hasExpTeam)   {
                Console.WriteLine("Нет опыта работы в команде.");
            }
            if (finalAmount > FundingAmount)
                Console.WriteLine($"Сумма {finalAmount} превышает финансирование {FundingAmount}.");
        }
    }
}

class InfrastructureProject : Project {
    public TimeSpan ExecutionPeriod {get; set;}

    public override void GetProjectDetails() {
        Console.WriteLine($"ID проекта: {ProjectId}");
        Console.WriteLine($"Название проекта: {ProjectName}");
        Console.WriteLine($"Статус проекта: {Status}");
        Console.WriteLine($"Тип: Инфраструктурный проект");
        Console.WriteLine($"Срок выполнения: {ExecutionPeriod.Days} дней");
    }
}

ProductProject myProductProject = new ProductProject();
myProductProject.ProjectId = 101;
myProductProject.ProjectName = "Мобильное приложение";
myProductProject.Status = "В разработке";
myProductProject.ExpectedCompletionDate = new DateTime(2026, 6, 30);
myProductProject.UpdateStatus("Активен");
Console.WriteLine();

ResearchProject myResearchProject = new ResearchProject();
myResearchProject.ProjectId = 202;
myResearchProject.ProjectName = "ИИ для медицины";
myResearchProject.Status = "Исследование";
myResearchProject.FundingAmount = 500_000m;

myResearchProject.AssignTeamMember(hasExpTeam: true, finalAmount: 100_000m, memberName: "Петров П.П.", hasExp: true);
Console.WriteLine();

myResearchProject.AssignTeamMember(hasExpTeam: true, finalAmount: 100_000m, memberName: "Сидоров С.С.", hasExp: false);
Console.WriteLine();

InfrastructureProject myInfrastructureProject = new InfrastructureProject();
myInfrastructureProject.ProjectId = 303;
myInfrastructureProject.ProjectName = "Дата-центр";
myInfrastructureProject.Status = "Планирование";
myInfrastructureProject.ExecutionPeriod = TimeSpan.FromDays(365);
myInfrastructureProject.GetProjectDetails();
```
